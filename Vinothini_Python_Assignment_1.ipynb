```python
# 1. STRING CONCATENATION
print("===== 1. STRING CONCATENATION =====")
string1 = "Hello "
name = input("Enter your Name: ")
string2 = string1 + name
print(string2)
string3 = ", welcome to Python programming"
string2 = string2 + string3
print(string2)

#2. STRING SLICING AND INDEXING
print("====== 2. STRING SLICING AND INDEXING ======")
text = string2
print("a. First character:", text[0])
print("b. Last character:", text[-1])
print("c. First 5 characters:", text[:5])
print("d. Last 11 characters:", text[-11:])
print("e. Reverse:", text[::-1])
python_start = text.index("Python")
print("f. Python:", text[python_start:python_start + 6])

# 3. STRING METHODS
print("====== 3. STRING METHODS ======")
strM = "Python beginner tutorial"
print("a. Uppercase:", strM.upper())
print("b. Lowercase:", strM.lower())
print("c. Capitalize:", strM.capitalize())
print("d. Number of 't':", strM.count('t'))
print("e. Replace 'Python' with 'Machine Learning':", strM.replace('Python', 'Machine Learning'))

# 4. TUPLES
print("====== 4. TUPLES ======")
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
t_combine = tuple1 + tuple2
print("a. Combined tuple:", t_combine)
t_repeat = t_combine * 3
print("b. Repeated tuple:", t_repeat)
print("c. 3rd element:", t_combine[2])
print("d. First Three elements:", t_combine[0:3])
print("e. Last Three elements:", t_combine[-4:-1])
```
